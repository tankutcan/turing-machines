# Random Turing Machines

```python
import matplotlib.pyplot as plt
import numpy as np
import random
```

```python
n = 4;
TM = {}
for k in range(n):
    _dict = {}
    for i in [0,1]:
        binary_array = [random.randint(0, 1) for _ in range(2)] +[random.randint(1,n)]
        _dict[i]= binary_array


    TM[k+1] = _dict
```

```python
def make_TM(n):
    TM = {}
    for k in range(n):
        _dict = {}
        for i in [0,1]:
            binary_array = [random.randint(0, 1) for _ in range(2)] +[random.randint(1,n)]
            _dict[i]= binary_array


        TM[k+1] = _dict

    return TM
```

```python

n = 4;
TM = make_TM(n)
```

```python
TM
```

```python
L = 1000;

tape = np.array([random.randint(0, 1) for _ in range(L)])#np.zeros(L)
current_position = int(L/2)
current_state = 1;

spacetime_tape = [tape]
position_trace = [current_position]
state_trace = [current_state]

arr = np.zeros(n)
current_value = tape[current_position]
arr[current_state-1] = 2*current_value - 1

current_tape = [arr]
for i in range(1000):
    current_value = tape[current_position]
    write, move, new_state = TM[current_state][current_value]
    arr = np.zeros(n)
    arr[current_state-1] = 2*current_value - 1
    
    tape[current_position] = write
    current_position = current_position + 2*move - 1
    current_state = new_state

    spacetime_tape +=[ tape.copy()]
    position_trace +=[current_position]
    state_trace +=[current_state]
    
    current_tape +=[arr]
```

```python
plt.imshow(current_tape,aspect = 'auto')
plt.show()
```
